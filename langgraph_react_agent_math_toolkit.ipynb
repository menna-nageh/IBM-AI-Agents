```python
from langchain_core.tools import tool
```

```python
@tool
def sum_numbers(numbers: list[float]) -> float:
    """Add a list of numbers together."""
    return sum(numbers)
```

```python
result = sum_numbers.invoke({
    "numbers": [-10, -20, -30]
})

print(result)
```

```python
# Math Toolkit
@tool
def add(a: float, b: float) -> float:
    """Add two numbers."""
    return a + b
```

```python
@tool
def subtract(a: float, b: float) -> float:
    """Subtract the second number from the first number."""
    return a - b
```

```python
@tool
def multiply(a: float, b: float) -> float:
    """Multiply two numbers."""
    return a * b
```

```python
@tool
def divide(a: float, b: float) -> float:
    """Divide the first number by the second number."""
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    return a / b
```

```python
print(add.invoke({"a": 10, "b": 5}))
print(subtract.invoke({"a": 10, "b": 5}))
print(multiply.invoke({"a": 10, "b": 5}))
print(divide.invoke({"a": 10, "b": 5}))
```

```python
from langchain_community.utilities import WikipediaAPIWrapper
```

```python
wiki = WikipediaAPIWrapper()
```

```python
#Create Wikipedia Tool

@tool
def search_wikipedia(query: str) -> str:
    """Search Wikipedia for information about a given topic."""
    return wiki.run(query)
```

```python
result = search_wikipedia.invoke({
    "query": "What is tool calling?"
})

print(result)
```

```python
tools = [
    add,
    subtract,
    multiply,
    divide,
    search_wikipedia
]
```

```python
# Create the final Agent

agent = create_react_agent(
    model=llm,
    tools=tools,
    prompt="You are a helpful assistant. Use the available tools whenever necessary."
)
```
