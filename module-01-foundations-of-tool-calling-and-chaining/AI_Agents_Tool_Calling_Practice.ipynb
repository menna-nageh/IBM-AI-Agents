```python
# Create a simple tool
from langchain_core.tools import tool

@tool
def add_numbers(a: float, b: float) -> float:
    """Add two numbers together."""
    return a + b
```

```python
# Test the Tool WITHOUT an Agent
result = add_numbers.invoke({
    "a": 3,
    "b": 2
})

print(result)
```

# **Level 2 — Structured Tool**

```python
from langchain_core.tools import tool

@tool
def add_numbers_with_options(
    numbers: list[float],
    absolute: bool = False
) -> float:
    """Sum a list of numbers. If absolute is True, sum their absolute values."""

    if absolute:
        numbers = [abs(x) for x in numbers]

    return sum(numbers)
```

```python
result = add_numbers_with_options.invoke({
    "numbers": [-2, 3],
    "absolute": False
})

print(result)
```

```python
result = add_numbers_with_options.invoke({
    "numbers": [-2, 3],
    "absolute": True
})

print(result)
```
